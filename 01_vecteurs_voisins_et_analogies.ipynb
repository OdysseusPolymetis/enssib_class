# 01 — Voisins et analogies dans un espace vectoriel

```python
import gensim.downloader as api
model = api.load('glove-wiki-gigaword-50')
print('Modèle chargé :', len(model), 'mots')
```

## Étape 1 — voisins proches

On commence par regarder les mots les plus proches de quelques professions.

```python
for word in ['doctor', 'nurse', 'teacher', 'engineer', 'programmer']:
    print('', word.upper())
    for w,score in model.most_similar(word, topn=8):
        print(f'{w:18s} {score:.3f}')
```

## Étape 2 — analogies

On teste la forme classique : `A - B + C = ?`.

```python
tests = [
    (['woman', 'doctor'], ['man']),
    (['man', 'nurse'], ['woman']),
    (['woman', 'professor'], ['man']),
    (['man', 'homemaker'], ['woman']),
]
for pos, neg in tests:
    print(f"\npositive={pos}, negative={neg}")
    print(model.most_similar(positive=pos, negative=neg, topn=10))
```
